STEP 2 — Separate expression data

```python
# Separate gene IDs
gene_ids = df['gene']

# Expression data only
expr = df.drop(columns=['gene'])

print(expr.head())
```

```python
print(expr.describe())
```

```python
# Log transform
expr_log = np.log1p(expr)

print(expr_log.head())
```

```python
# Open SOFT metadata file

with open("GSE261878_family.soft", "r", encoding="utf-8") as f:
    soft_text = f.read()

# Preview first 3000 characters
print(soft_text[:3000])
```

```python
for line in soft_text.split("\n"):
    if "^SAMPLE" in line:
        print(line)

    if "!Sample_title" in line:
        print(line)

    if "!Sample_characteristics_ch1" in line:
        print(line)
```

```python
import re
import pandas as pd

samples = []
genotypes = []

current_sample = None

for line in soft_text.split("\n"):

    # Extract sample title
    if "!Sample_title" in line:

        match = re.search(r"\((.*?)\)", line)

        if match:
            current_sample = match.group(1)

    # Extract genotype
    if "genotype:" in line:

        genotype = line.split("genotype:")[-1].strip()

        samples.append(current_sample)
        genotypes.append(genotype)

# Create metadata dataframe
metadata = pd.DataFrame({
    "sample": samples,
    "label": genotypes
})

print(metadata.head())
print(metadata.shape)
```

```python
# Set gene IDs as row index BEFORE transpose

expr_indexed = expr_log.copy()

expr_indexed.index = gene_ids

# Transpose
X = expr_indexed.T

print(X.head())
print(X.shape)
```

```python
# Set sample names as index
metadata = metadata.set_index("sample")

# Match labels
X['label'] = metadata.loc[X.index, 'label']

print(X[['label']].head())
```

```python
# Features = all gene columns
X_features = X.drop(columns=['label'])

# Target = labels
y = X['label']

# Check
print(X_features.shape)

print("\nFirst 5 labels:")
print(y.head())
```

```python
#Splitting dataset(test, train)
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_features, y, test_size=0.2, random_state=42
)
```

```python
#training the Model
from sklearn.ensemble import RandomForestClassifier

rfc = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

rfc.fit(X_train, y_train)
```

```python
#Prediction
y_pred = rfc.predict(X_test)
```

```python
#Evaluation
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Accuracy:", accuracy_score(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
#Cross Validation
from sklearn.model_selection import cross_val_score

scores = cross_val_score(rfc, X_features, y, cv=5)
print(scores)
print("Mean CV accuracy:", scores.mean())
```

```python
#Using pipeline to avoid data leakage
from sklearn.pipeline import Pipeline
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.model_selection import StratifiedKFold, cross_val_score

pipeline = Pipeline([
    ('select', SelectKBest(f_classif, k=30)),
    ('model', rfc)
])

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

scores = cross_val_score(pipeline, X_features, y, cv=cv)

print(scores)
print(scores.mean())
```

```python
#removing constant features
from sklearn.feature_selection import VarianceThreshold

X_filtered = VarianceThreshold(threshold=0.0).fit_transform(X_features)
```

```python
#re=running pipeline
from sklearn.pipeline import Pipeline
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.model_selection import StratifiedKFold, cross_val_score

pipeline = Pipeline([
    ('var_filter', VarianceThreshold()),
    ('select', SelectKBest(f_classif, k=30)),
    ('model', rfc)
])

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

scores = cross_val_score(pipeline, X_features, y, cv=cv)

print(scores)
print(scores.mean())
```

```python
#Analyzing important genes
from sklearn.feature_selection import SelectKBest, f_classif

selector = SelectKBest(score_func=f_classif, k=30)
X_new = selector.fit_transform(X_features, y)

# Get selected gene names
selected_genes = X_features.columns[selector.get_support()]
print(selected_genes)
```

```python
#Converting gene IDs into rankable table
import pandas as pd

gene_scores = pd.DataFrame({
    'Gene': X_features.columns,
    'Score': selector.scores_
})

top_genes = gene_scores.sort_values(by='Score', ascending=False).head(30)
print(top_genes)
```
