1. Store the DNA fragment "GGATCCGATTA" in a variable and print it.

```python
dna_fragment = "GGATCCGATTA"
print(f"DNA fragment: {dna_fragment}")
```

2. Count how many "T" bases are present in the DNA fragment above.

```python
count_base = dna_fragment.count("T")
print(f"Count of 'T' base in DNA fragment: {count_base}")
```

3. Find the length of the DNA fragment.

```python
length_dna = len(dna_fragment)
print(f"Length of DNA fragment: {length_dna}")
```

4. Calculate the AT content (number of A and T bases divided by total length).

```python
A_content = dna_fragment.count("A")
T_content = dna_fragment.count("T")
AT_content = (A_content + T_content) / length_dna
print(f"AT content of dna_fragment: {AT_content}")
```
